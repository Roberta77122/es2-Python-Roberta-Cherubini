```python
from abc import ABC, abstractmethod

class Person(ABC):
    def __init__(self, nome, eta):
        self.nome = nome
        self.eta = eta 

    def presentazione(self):
        return f"Sono {self.nome} e ho {self.eta} anni."
    
    @abstractmethod
    def lavoro(self):
        pass


class Dipendente(Person):

    def __init__(self, nome, eta, stipendio):
        super().__init__(nome, eta)
        self.stipendio = stipendio

    @abstractmethod
    def lavoro(self):
        pass 

class Studente(Person):

    def __init__(self, nome, eta, materia):
        super().__init__(nome, eta)
        self.materia = materia

    def lavoro(self):
        return f"Sono uno studente di {self.materia}!"  

class Docente(Dipendente):
    def __init__(self, nome, eta, stipendio, azienda):
        super().__init__(nome, eta, stipendio)
        self.azienda= azienda
    def lavoro(self):
            return f"Sono un docente di {self.azienda}!"  


class Freelance(Dipendente):   
    def __init__(self, nome, eta, stipendio, specializzazione):
        super().__init__(nome, eta, stipendio)
        self.specializzazione = specializzazione
    def lavoro(self):
            return f"Sono un Freelance specializzato in {self.specializzazione}!"  


studente = Studente("Luca", 22, "Python")
docente = Docente("Marco", 33, 9999, "Aulab")
freelance = Freelance("Anna", 29, 3000, "Data Science")

print(docente.presentazione())
print(docente.lavoro())



```
